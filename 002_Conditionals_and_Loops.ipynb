# **if else Condition**

**Find number is negative, positive or zero**

```python
%%writefile condition1.cpp

#include<iostream>
using namespace std;
int main(){
  int n;
  cin >> n;
  if (n>0){
    cout << "positive";
  }
  else if (n<0){
    cout << "negative";
  }
  else{
    cout << "zero";
  }
}
```

```python
!g++ condition1.cpp -o condition1
!./condition1
```

**Print "lowwer case" if char is between a-z, print "upper case" if char is between A-Z, else in between 0-9 print "numeric"**

```python
%%writefile condition3.cpp

#include<iostream>
using namespace std;
int main(){
  char c;
  cout<<"Enter a character: ";
  cin>>c;
  if (c>='a' && c<='z'){
    cout<<"lower case";
  }
  else if (c>='A' && c<='Z'){
    cout<<"upper case";
  }
  else{
    cout<<"numeric";
  }
}
```

```python
!g++ condition3.cpp -o condition3
!./condition3
```

# **Loops**

**Print number from 1 to n**

```python
%%writefile loop1.cpp

#include<iostream>
using namespace std;
int main(){
  int n;
  cin >>n;
  int i=1;
  while (i<=n){
    cout << i << " ";
    i++;
  }
}
```

```python
!g++ loop1.cpp -o loop1
!./loop1
```

**Sum of number from 1 to n**

```python
%%writefile loop2.cpp

#include<iostream>
using namespace std;
int main(){
  int n;
  int sum = 0;
  cin >>n;
  int i=1;
  while (i<=n){
    sum+=i;
    i++;
  }
  cout<<sum;
}
```

```python
!g++ loop2.cpp -o loop2
!./loop2
```

**Sum of even number from 1 to n**

```python
%%writefile loop3.cpp

#include<iostream>
using namespace std;
int main(){
  int n;
  cin >>n;
  int sum = 0;
  int i=1;
  while (i<=n){
    if (i%2==0){
      sum+=i;
    }
    i++;
  }
  cout<<sum;
}
```

```python
!g++ loop3.cpp -o loop3
!./loop3
```

**Check number n is prime or not**

```python
%%writefile loop5.cpp

#include<iostream>
using namespace std;
int main(){
  int n;
  cin >>n;
  int i=2;
  while (i<n){
    if (n%i==0){
      cout<<"not prime";
      return 0;
    }
    i++;
  }
  cout<<"prime";
}
```

```python
!g++ loop5.cpp -o loop5
!./loop5
```
