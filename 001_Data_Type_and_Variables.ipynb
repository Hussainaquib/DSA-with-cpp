**Data Type**

```python
%%writefile datatype.cpp

#include<iostream>
using namespace std;
int main(){
    int a = 123;
    cout<<a<<endl;
    int size1 = sizeof(a);
    cout<<size1<<endl;

    char b = 'v';
    cout<<b<<endl;
    int size2 = sizeof(b);
    cout<<size2<<endl;

    bool bl = true;
    cout<<bl<<endl;
    int size3 = sizeof(bl);
    cout<<size3<<endl;

    float f = 1.2;
    cout<<f<<endl;
    int size4 = sizeof(f);
    cout<<size4<<endl;

    double d = 1.234567;
    cout<<d<<endl;
    int size5 = sizeof(d);
    cout<<size5<<endl;
}
```

```python
! g++ datatype.cpp -o datatype
! ./datatype
```

**Type Casting**

```python
%%writefile typecast.cpp

#include<iostream>
using namespace std;
int main(){
    int a = 'a';
    cout<<a<<endl;

    char ch = 98;
    cout<<ch<<endl;

    return 0;
}
```

```python
! g++ typecast.cpp -o typecast
! ./typecast
```

**Unsigned Int**

```python
%%writefile unsigned.cpp

#include<iostream>
using namespace std;
int main(){
    unsigned int a1 = 112;
    cout<<a1<<endl;

    unsigned int a2 = -112;
    cout<<a2<<endl;

    return 0;
}
```

```python
! g++ unsigned.cpp -o unsigned
! ./unsigned
```
